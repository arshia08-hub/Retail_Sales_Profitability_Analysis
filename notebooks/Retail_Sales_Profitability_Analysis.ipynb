```python
# Q1.How much Business are we doing???
# Ans: calculate total sales and profit(sales only can't tell whether the business is goign well)
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
```

```python
#calculate profit 
profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", profit_margin)
```

```python
# Q2.Which product category brings in the most sales, and which actually makes the most money? (sort by total profit)
category_analysis = df.groupby("product_sub-category").agg(
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum")
).sort_values("total_profit", ascending=False)

category_analysis
```

```python
# Calculate profit margin by category
category_analysis["profit_margin"] = (
    category_analysis["total_profit"]
    / category_analysis["total_sales"]
) * 100
category_analysis
```

```python
# Now we have:category | total_sales | total_profit | profit_margin
```

```python
# Region analysis
region_analysis = df.groupby("region").agg(
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum")
).sort_values("total_profit", ascending=False)

region_analysis
```

```python
region_analysis["profit_margin"] = (
    region_analysis["total_profit"]
    / region_analysis["total_sales"]
) * 100

region_analysis
```
